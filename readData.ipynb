```python
from scipy.io import loadmat
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
### Reading only the first file as of now. breaks after that. 
path = 'Data Session 1/david_20200123_session1_0'
ext = '.mat'
data = pd.DataFrame()
for i in range(1, 9):
    filename = path + str(i) + ext
    file_data = loadmat(filename)
    mdata = file_data['data']
    temp = pd.DataFrame(mdata)
    data = pd.concat([data, temp], sort=False)
    break
```
