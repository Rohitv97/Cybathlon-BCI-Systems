```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from scipy.io import loadmat

%matplotlib inline
```

```python
### Reading all files
path = 'data/david_20200123_session1_0'
ext = '.mat'
data = pd.DataFrame()
for i in range(1, 9):
    filename = path + str(i) + ext
    file_data = loadmat(filename)
    mdata = file_data['data']
    temp = pd.DataFrame(mdata)
    data = pd.concat([data, temp])
data.shape
```

```python
cols = np.arange(73)
columns = []
[columns.append(str(x)) for x in cols]
columns
```

```python
data = data.values
df = pd.DataFrame(data, columns=columns)
df
```

```python
df = df.drop(['67','68','69','70','71'], axis=1)
df.head()
```

```python
columns.remove('67')
columns.remove('68')
columns.remove('69')
columns.remove('70')
columns.remove('71')
```

```python
sfreq = 512
trigger_points = {1:2, 2:2, 3:2, 4:2 }
asynch_label = []
i = 0
while i < len(df['72']):
    if(df['72'][i] == 1):
        end = (sfreq * trigger_points[1]) + 1
        for j in range(i, i+end):
            asynch_label.append(1)
        i += end
    elif(df['72'][i] == 2):
        end = (sfreq * trigger_points[2]) + 1
        for j in range(i, i+end):
            asynch_label.append(2)
        i += end
    elif(df['72'][i] == 3):
        end = (sfreq * trigger_points[3]) + 1
        for j in range(i, i+end):
            asynch_label.append(3)
        i += end
    elif(df['72'][i] == 4):
        end = (sfreq * trigger_points[4]) + 1
        for j in range(i, i+end):
            asynch_label.append(4)
        i += end
    elif(df['72'][i] == 0):
        asynch_label.append(0)
        i += 1
```

```python
df['asynch_label'] = asynch_label
```

```python
asynch_label = np.array(asynch_label)
```

```python
def data_win(sfreq, data, asynch_label):
    sampling_window = 2 * sfreq
    shift_length = 1 * sfreq
    t_start = 0
    
    new_data = []
    labels = []


    while t_start + sampling_window < data.shape[0]:
        new_data.append(data.iloc[t_start:t_start+sampling_window, :64].T.values)
        labels.append(asynch_label[t_start:t_start+sampling_window])
        #labels.append(data.iloc[t_start:t_start+sampling_window, -1].values)
    
        t_start = t_start + shift_length
    
    return np.array(new_data), np.array(labels)
```

```python
data_new, label_new = data_win(sfreq, df, asynch_label)
```

```python
#### Converting label to shape (2591, 1)
label = []

for i in label_new:
    count1 = np.count_nonzero(i==1)
    count4 = np.count_nonzero(i==4)
    if count1 > 512:
        to_add = 1
    elif count4 > 512:
        to_add = 4
    else:
        to_add = 0
    label.append(to_add)
    
label = np.array(label)
```

```python
data_new.shape

train_data = []
label_data = []

for i in range(len(label)):
    if label[i] != 0:
        label_data.append(label[i])
        train_data.append(data_new[i])
label_data = np.array(label_data)
train_data = np.array(train_data)
```

```python
print(label_data.shape)
print(train_data.shape)
```

```python
event_dict = {'L_Hands': 1, 'M_Hands': 2, 'S_Hands': 3, 'Relax': 4}
```

```python
import mne
from mne.decoding import CSP
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(train_data, label_data, test_size=0.2, random_state=42)
```

```python
csp = CSP(n_components=4, reg=None, log=True, norm_trace=False)
```

```python
X_train_csp = csp.fit_transform(X_train, y_train)
```

```python
X_test_csp = csp.transform(X_test)
```

```python
from sklearn.svm import SVC
svm = SVC()
svm.fit(X_train_csp, y_train)
```

```python
from sklearn.metrics import accuracy_score, f1_score, cohen_kappa_score
y_pred = svm.predict(X_test_csp)
print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Cohen Kappa: ', cohen_kappa_score(y_test, y_pred))
```
