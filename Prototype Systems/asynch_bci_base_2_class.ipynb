```python
sfreq = 512
trigger_points = {1:6, 2:4, 3:2, 4:2 }
asynch_label = []
real_label = []
i = 0
while i < len(df['72']):
    if(df['72'][i] == 1):
        end = (sfreq * trigger_points[1]) + 1
        for j in range(i, i+end):
            real_label.append(1)
            asynch_label.append(1)
        i += end
    elif(df['72'][i] == 2):
        end = (sfreq * trigger_points[2]) + 1
        for j in range(i, i+end):
            real_label.append(2)
            asynch_label.append(1)
        i += end
    elif(df['72'][i] == 3):
        end = (sfreq * trigger_points[3]) + 1
        for j in range(i, i+end):
            real_label.append(3)
            asynch_label.append(1)
        i += end
    elif(df['72'][i] == 4):
        end = (sfreq * trigger_points[4]) + 1
        for j in range(i, i+end):
            real_label.append(4)
            asynch_label.append(4)
        i += end
    elif(df['72'][i] == 0):
        real_label.append(0)
        asynch_label.append(0)
        i += 1
```

```python
def data_win(sfreq, data, asynch_label, real_label):
    sampling_window = 2 * sfreq
    shift_length = 1 * sfreq
    t_start = 0
    
    new_data = []
    labels = []
    real = []


    while t_start + sampling_window < data.shape[0]:
        new_data.append(data.iloc[t_start:t_start+sampling_window, :64].T.values)
        labels.append(asynch_label[t_start:t_start+sampling_window])
        real.append(real_label[t_start:t_start+sampling_window])
        #labels.append(data.iloc[t_start:t_start+sampling_window, -1].values)
    
        t_start = t_start + shift_length
    
    return np.array(new_data), np.array(labels), np.array(real)
```

```python
data_new, label_new, real = data_win(sfreq, df, asynch_label, real_label)
```

```python
#### Converting label to shape (2591, 1)
label = []

for i in label_new:
    count1 = np.count_nonzero(i==1)
    count4 = np.count_nonzero(i==4)
    if count1 >= 512:
        to_add = 1
    elif count4 >= 512:
        to_add = 4
    else:
        to_add = 0
    label.append(to_add)
    
label = np.array(label)

label_4 = []
for i in real:
    count1 = np.count_nonzero(i==1)
    count2 = np.count_nonzero(i==2)
    count3 = np.count_nonzero(i==3)
    count4 = np.count_nonzero(i==4)
    if count1 >= 512:
        to_add = 1
    elif count2 >= 512:
        to_add = 2
    elif count3 >= 512:
        to_add = 3
    elif count4 > 512:
        to_add = 4
    else:
        to_add = 0
    label_4.append(to_add)
    
label_4= np.array(label_4)
```
